```python
#Create a simple decorator that adds extra print statements before and after a function execution


def greet(fx):
    def mfx():
        print("Good Morning")
        fx()
        print("Thank uh for using this function")   
    return mfx

@greet
def hello():
    print("hello world")

hello()
```

```python
def greet(fx):
    def mfx(*args,**kwargs):
        print("Good Morning")
        fx(*args,**kwargs)
        print("Thank u for using this function")
    return mfx  

def add(a,b):
    print(a+b)

greet(add)(2,4)     
```

```python
#Create a decorator that wraps a function and prints a custom message before calling the original function

def my_decorator(fx):
    def wrap():
        print("Thank you for using this function")
        fx()
        
    return wrap

@my_decorator
def say_hello():
    print("All The Best")

say_hello()
```

```python
#Create a decorator to measure execution time of a function that takes no arguments

import time 

def time_decorator(func):
    def wrap():
        start=time.time()
        result=func()
        end=time.time()
        print(f"time taken:{end - start:.4f} second")
        return result
    return wrap

@time_decorator
def work_time():
    time.sleep(1)
    print("work done. ")

work_time()
```

```python
#Create a decorator that checks if the input user is "admin" before allowing function execution

def requires_admin(func):
    def wrapper(user):
        if user != "admin":
            print("Access denied")
            return
        return func(user)
    return wrapper

@requires_admin
def access(user):
    print(f"{user} accessed the dashboard")

access("guest")
access("admin")

```

```python
#Create a decorator that works with functions having arguments using *args and **kwargs and prints messages before and after execution

def log_decorator(func):
    def wrapper(*args,**kwargs):
        print(f"➡️ Starting {func.__name__} with args={args}, kwargs={kwargs}")

        result=func(*args,**kwargs)
        print(f"✅ Finished {func.__name__}, result={result}")
        return result
    return wrapper

@log_decorator
def multiply(a,b):
    return a*b

print(multiply(4,2))
```

```python
#Create a decorator that modifies the return value of a function by doubling the result

def double_result(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        doubled = result * 2
        print(f"🔄 {func.__name__} returned {result}, doubled → {doubled}")
        return doubled
    return wrapper

@double_result
def add(a, b):
    return a + b

@double_result
def square(n):
    return n * n

print(add(3, 4))     
print(square(5))    
```
