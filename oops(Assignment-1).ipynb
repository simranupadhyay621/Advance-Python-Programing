```python
# Question 1: Laptop class
class Laptop:
    def __init__(self, price):
        self.__price = price

    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price

    def get_price(self):
        return self.__price
laptop = Laptop(50000)
print(laptop.get_price())

laptop.set_price(60000)
print(laptop.get_price())

laptop.set_price(-10000)
print(laptop.get_price())
```

```python
# Question 2: BankAccount class
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0: self.__balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.__balance: self.__balance -= amount

    def get_balance(self):
        return self.__balance

account = BankAccount(1000)
print(account.get_balance())

account.deposit(500)
print(account.get_balance())

account.withdraw(200)
print(account.get_balance())

account.withdraw(2000)
print(account.get_balance())
```

```python
# Question 3: Movie class
class Movie:
    def __init__(self, rating):
        self.__rating = max(1, min(10, rating))

    def set_rating(self, new_rating):
        if 1 <= new_rating <= 10:
            self.__rating = new_rating

    def get_rating(self):
        return self.__rating

movie = Movie(8)
print(movie.get_rating())

movie.set_rating(10)
print(movie.get_rating())

movie.set_rating(15)
print(movie.get_rating())

movie2 = Movie(-3)
print(movie2.get_rating())

```

```python
# Question 4: Book class
class Book:
    def __init__(self, title, author):
        self.__title = title
        self.__author = author

    def get_title(self):
        return self.__title

    def get_author(self):
        return self.__author

book = Book("1984", "George Orwell")
print(book.get_title())   # Output: 1984
print(book.get_author())  # Output: George Orwell
```

```python
# Question 5: Car class
class Car:
    def __init__(self, speed=0):
        self.__speed = speed

    def increase_speed(self, amount):
        if amount > 0: self.__speed += amount

    def decrease_speed(self, amount):
        if 0 < amount <= self.__speed: self.__speed -= amount

    def get_speed(self):
        return self.__speed

car = Car(50)              # Start at 50 km/h
print(car.get_speed())     # Output: 50

car.increase_speed(20)     # Increase by 20
print(car.get_speed())     # Output: 70

car.decrease_speed(30)     # Decrease by 30
print(car.get_speed())     # Output: 40

car.decrease_speed(100)    # Invalid (too much), ignored
print(car.get_speed())     # Output: 40
```

```python
#Create an ElectricCar class that inherits from the Car class and includes an additional attribute, battery_size.
class Car:
    def __init__(self, model,brand):
        self.model = model
        self.brand = brand

    def time(self):
        return "refuling time is 2 hour"
class type(Car):
    def __init__(self,model,brand,battery_size):
        super().__init__(model,brand)
        self.battery_size=battery_size

    def time(self):
        return "charging time is 2 hour"
car1=type("Fortuner","Toyota","80KW")
print(car1.model,car1.brand,car1.battery_size)
```

```python
#Create a Laptop class that inherits from a Computer class and includes an additional attribute battery_life
class laptop:
    def __init__(self,brand,proccessor):
        self.brand=brand
        self.proccessor=proccessor
class battery(laptop):
    def __init__(self,brand,proccessor,battery_life):
        super().__init__(brand,proccessor)
        self.battery_life=battery_life
s1=battery("Lenovo","H-type","6hours")
print(s1.brand,s1.proccessor,s1.battery_life)
```

```python
#Define a Smartphone class and add another child class of your choice that inherits from an ElectronicDevice class
#and includes an additional attribute operating_system
class smartphone():
    def __init__(self,brand,storage):
        self.brand=brand
        self.storage=storage
class device(smartphone):
    def __init__(self,brand,storage,os):
        super().__init__(brand,storage)
        self.os=os
x=device("Samsung Galaxy S23 Ultra","12gb","Android 13")
print(x.brand,x.storage,x.os)
```

```python
#Define a Musician class and add another child class of your choice that extends an Artist class and includes an
#additional attribute instrument_played
class musician():
    def __init__(self,name,age):
        self.name=name
        self.age=age
class artist(musician):
    def __init__(self,name,age,play):
        super().__init__(name,age)
        self.play=play
x= artist("Zakhir Hussain",73,"Tabla")
print(x.name,x.age,x.play)
x2= artist("A.R Rahman",58,"Piano")
print(x2.name,x2.age,x2.play)
```

```python
#Implement a FictionBook class, add another child class of your choice that inherits from a Book class and add an attribute genre
class fictionbook:
    def __init__(self,writer,name):
        self.writer=writer
        self.name=name
class book(fictionbook):
    def __init__(self,writer,name,genre):
        super().__init__(writer,name)
        self.genre=genre
s1=book("SPY×FAMILY1","Tatsuya Endo","slice-of-life")
print(s1.writer,s1.name,s1.genre)
```
